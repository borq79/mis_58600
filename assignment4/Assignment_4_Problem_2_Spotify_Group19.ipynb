Find all songs that satisfy both conditions:

## Part C: Transforming and Creating New Columns (15 Points)

### C.1
Create a new column called `duration_min` that expresses `duration_ms` in minutes, rounded to two decimal places.

```python
# C.1
spotify['duration_min'] = round(spotify['duration_ms'] / 60000, 2)
display(spotify.head(5))
```

### C.2
Create a new column called `energy_level` using `map()`. First, use conditional logic to assign each song an integer category based on its energy:
- 0 for energy below 0.40
- 1 for energy from 0.40 up to, but not including, 0.75
- 2 for energy of 0.75 or higher

Then use `map()` with a dictionary to convert these categories into "Low", "Moderate", and "High".

```python
# C.2
song_energy_map = {
    0: "Low",
    1: "Moderate",
    2: "High"
}

energy_range = [
    spotify['energy'] < 0.40,
    ((spotify['energy'] >= 0.40) & (spotify['energy'] < 0.75)),
    spotify['energy'] >= 0.75]


spotify['energy_range'] = np.select(energy_range, [0, 1, 2])
spotify['energy_level'] = spotify['energy_range'].map(song_energy_map)
```

### C.3
Create a column called `workout_candidate` using `apply()` and a lambda function. Assign "Yes" to songs with energy > 0.75 **and** danceability > 0.65; assign "No" to all other songs.

```python
# C.3
spotify['workout_candidate'] = spotify.apply( lambda s: "Yes" if ((s['energy'] > 0.75) & (s['danceability'] > 0.65)) else "No", axis=1 )
```

### C.4
Calculate the total number and percentage of songs classified as workout candidates.

```python
# C.4
num_of_rows = len(spotify)
num_of_wc   = len(spotify[spotify['workout_candidate'] == "Yes"])
print("Total Number of Songs                          :",  num_of_rows)
print("Total Number of Workout Candidates             :",  num_of_wc)
print(f"Percentage of Songs that are Workout Candidates: {round( (num_of_wc/num_of_rows)*100, 2)}%")
```

## Part D: Data Cleaning and Quality Checks (10 Points)

### D.1
Identify the number of missing values in `track_name`, `track_artist`, `track_popularity`, and `energy`.

```python
# D.1
num_of_missing_track_name = spotify['track_name'].isna().sum()
num_of_missing_track_artist = spotify['track_artist'].isna().sum()
num_of_missing_track_popularity = spotify['track_popularity'].isna().sum()
num_of_missing_energy = spotify['energy'].isna().sum()

print("Number of Tracks missing Track Name       : ", num_of_missing_track_name)
print("Number of Tracks missing Track Artist     : ", num_of_missing_track_artist)
print("Number of Tracks missing Track Popularity : ", num_of_missing_track_popularity)
print("Number of Tracks missing Energy           : ", num_of_missing_energy)
```

### D.2
Remove rows with missing `track_name` or `track_artist` values.

```python
# D.2
print("Number of songs prior to dropping NaN data :", spotify.shape[0])
spotify = spotify.dropna(subset=['track_name', 'track_artist'])
print("Number of songs after dropping NaN data :", spotify.shape[0])
```

### D.3
Check how many rows have duplicate `track_id` values. Since a track may appear in multiple playlists, explain why duplicate track IDs do not necessarily indicate incorrect data.

```python
# D.3
num_of_duplicates = spotify['track_id'].duplicated().sum()
print("Number of rows with duplicate 'track_id' values :", num_of_duplicates)
```

**Explanation:** 
Track id can be used to determine the specific song, however the entire row must be considered. For example, rows for song 'We Own It (Fast & Furious)', which track id '00QyLmjxaSEE8qIZQjBXBj', it is labeled with three different genres (rap, r&b, and edm) and is part of three different playlists. Therefore, they are not duplicates and should not be dropped or it could skew the data analysis results. They should be considered in their entirety, not just their track id. 

### D.4
Create a separate DataFrame containing one record per unique `track_id`, keeping the first occurrence. Report the number of records before and after removing duplicate track IDs.

_Note: use the original playlist-level DataFrame for Part E so that genre-specific information is retained._

```python
# D.4
print("Number of songs prior to dropping duplicate track_ids :", len(spotify))
spotify_unique_tracks = spotify.drop_duplicates(subset="track_id", keep="first")
print("Number of songs after dropping duplicate track_ids :", len(spotify_unique_tracks))
```

## Part E: Grouping, Aggregation, and Business Insights (10 Points)

### E.1
Group the songs by `playlist_genre` and calculate:
- Number of song records
- Average track popularity
- Average energy
- Average danceability

```python
# E.1
genre_summary = spotify.groupby("playlist_genre").agg(
    song_count=("track_id", "count"),
    avg_popularity=("track_popularity", "mean"),
    avg_energy=("energy", "mean"),
    avg_danceability=("danceability", "mean"),
).round(2)

genre_summary
```

### E.2
Sort the results by average track popularity in descending order.

```python
# E.2
genre_sorted = genre_summary.sort_values('avg_popularity', ascending=False)
print("Results sorted by average track popularity in descending order:")
display(genre_sorted)
```

### E.3
Identify the three playlist genres with the highest average energy.

```python
# E.3
top_energy_genres = genre_summary.nlargest(3, 'avg_energy')

print("The three playlist genres with the highest average energy:\n")
display(top_energy_genres[['avg_energy', 'avg_popularity', 'avg_danceability']])
```

### E.4
Group the songs by `playlist_genre` and `energy_level` to calculate the number of songs in each combination. Which genre contains the largest number of high-energy song records?

```python
# E.4
genre_energy_counts = (
    spotify.groupby(['playlist_genre', 'energy_level'])
         .size()
)

print("Number of songs in each genre\n")
display(genre_energy_counts)
```

**Answer:** The genre with the largest number of high-energy songs records, with 4,192 records, is edm.
